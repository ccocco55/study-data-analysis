```python
import pandas as pd

traveling_df = pd.read_csv('./datasets/내국인출국성별_20251120101412.csv', encoding='utf-8')
display(traveling_df)
```

```python
traveling_df = traveling_df[traveling_df['성별(1)'] != '승무원']
traveling_df
```

```python
traveling_df.reset_index(inplace=True)
traveling_df
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.lineplot(data=traveling_df, x='시점', y='데이터', hue='성별(1)', marker='o')

plt.xlabel('시점')
plt.ylabel('데이터 값')
plt.title('시점별 성별 데이터 변화')
plt.xticks(rotation=45)

plt.show()
```
